

# 1. Difference between AI, ML and Deep Learning



Artificial Intelligence (AI) is a broad field of computer science that focuses on creating machines that can perform tasks that normally require human intelligence. These tasks include decision-making, problem-solving, speech recognition, and language translation. For example, virtual assistants like Siri or Google Assistant are applications of AI.

Machine Learning (ML) is a subset of AI. It allows machines to learn from data and improve their performance without being explicitly programmed. Instead of writing fixed rules, we train the system using data. For example, an email spam filter learns to detect spam emails based on past examples.

Deep Learning is a subset of Machine Learning. It uses neural networks with many layers to learn complex patterns from large amounts of data. Deep Learning is mainly used in image recognition, speech recognition, and self-driving cars. For example, facial recognition systems in smartphones use deep learning.

# 3.What is an activation function? Explain the purpose of activation functions and describe Sigmoid, ReLU, and Tanh with their advantages and disadvantages.

An activation function decides whether a neuron should be activated or not. It adds non-linearity to the model, which helps the network learn complex patterns. Without activation functions, the neural network would behave like a simple linear model.

The Sigmoid function gives output values between 0 and 1. It is useful for binary classification but suffers from vanishing gradient problems. ReLU (Rectified Linear Unit) gives output as zero for negative values and keeps positive values unchanged. It is simple and fast but can suffer from the dying ReLU problem. Tanh gives outputs between -1 and 1. It works better than sigmoid in some cases but also suffers from vanishing gradients.

# 4. Explain the architecture of a basic Artificial Neural Network (ANN). Describe input layer, hidden layer, output layer, weights, bias, and forward propagation.

A basic Artificial Neural Network consists of three main layers: input layer, hidden layer, and output layer. The input layer receives the data. The hidden layer performs computations using weights, bias, and activation functions. The output layer produces the final result.

Each connection between neurons has a weight that shows the importance of the input. The bias helps adjust the output. Forward propagation is the process where input data moves from the input layer to the output layer through hidden layers, and predictions are calculated step by step.

# 5. What is backpropagation? Explain the role of the chain rule in updating weights. Why is gradient descent used?

Backpropagation is a method used to train neural networks. It calculates the error between the predicted output and the actual output. Then it adjusts the weights to reduce this error.

The chain rule from calculus is used to compute gradients of the loss function with respect to each weight. This helps in finding how much each weight contributes to the error. Gradient descent is used to update the weights in small steps so that the error gradually decreases and the model improves.

# 6. Differentiate between Batch Gradient Descent, Stochastic Gradient Descent (SGD), and Mini-batch Gradient Descent.

Batch Gradient Descent uses the entire training dataset to update the weights. It is stable but slow for large datasets. Stochastic Gradient Descent (SGD) updates the weights using only one training example at a time. It is faster but can be noisy and unstable. Mini-batch Gradient Descent is a combination of both. It updates weights using a small batch of data, making it faster and more stable. It is the most commonly used method.

# 7. What is overfitting in neural networks? Explain at least three techniques used to prevent overfitting.

Overfitting occurs when a neural network learns the training data too well, including noise and unwanted details. As a result, it performs well on training data but poorly on new data.

To prevent overfitting, we can use techniques like dropout, where some neurons are randomly turned off during training. Another method is regularization, which adds a penalty to large weights. Early stopping is also used, where training is stopped when validation error starts increasing.

# 8. Explain what a Convolutional Neural Network (CNN) is. Why are CNNs preferred for image data?

A Convolutional Neural Network is a special type of neural network mainly used for image data. It automatically detects important features such as edges, shapes, and patterns in images.

CNNs are preferred for image data because they use filters that focus on small regions of the image at a time. This reduces the number of parameters and makes the model more efficient. They also maintain spatial relationships between pixels.

# 9. Describe the following components of CNNs: (a) Convolution Layer (b) Filter/Kernel (c) Stride (d) Padding (e) Pooling Layer.

**(a) Convolution Layer**
The convolution layer is the main building block of a CNN. It is responsible for extracting important features from the input image, such as edges, textures, and patterns. Instead of looking at the whole image at once, it focuses on small parts of the image and processes them step by step. It performs a mathematical operation called convolution, where a small matrix (filter) slides over the image and calculates weighted sums. The result of this operation is called a feature map.

**(b) Filter / Kernel**
A filter, also called a kernel, is a small matrix of numbers that moves over the input image. Its size is usually smaller than the image, such as 3×3 or 5×5. The filter multiplies its values with the corresponding values of the image and adds them together to produce a single value. This process is repeated across the entire image. Different filters detect different features, such as vertical edges, horizontal edges, or specific textures.

**(c) Stride**
Stride refers to the number of pixels by which the filter moves over the image at each step. If the stride is 1, the filter moves one pixel at a time. If the stride is 2, it moves two pixels at a time. A larger stride reduces the size of the output feature map because the filter skips more pixels. Smaller stride keeps more detailed information but increases computation. Stride controls both output size and computational cost.

 **(d) Padding**
 Padding means adding extra pixels (usually zeros) around the border of the input image before applying convolution. This is done to control the size of the output feature map. Without padding, the output size becomes smaller after each convolution. With padding, we can preserve the original size of the image.

**(e) Pooling Layer**
The pooling layer is used to reduce the size of the feature map while keeping the most important information. It helps in reducing computation and preventing overfitting. The most common type is Max Pooling, where the maximum value is selected from a small region. Another type is Average Pooling, where the average value is taken.
